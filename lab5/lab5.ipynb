```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the image
image = cv2.imread('sample_image.png')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)


# Task 1: Convolve an image with a 7x7 box filter


# Create a 7x7 box filter (averaging filter)
kernel_7x7 = np.ones((7, 7), np.float32) / 49

# Apply the box filter
smoothed_box = cv2.filter2D(image, -1, kernel_7x7)

# Plot original and smoothed images side by side
fig1, ax1 = plt.subplots(1, 2, figsize=(12, 5))
fig1.suptitle("Task 1: 7x7 Box Filter", fontsize=16, fontweight='bold')

ax1[0].imshow(image)
ax1[0].set_title("Original Image")
ax1[0].axis("off")

ax1[1].imshow(smoothed_box)
ax1[1].set_title("Smoothed (7x7 Box Filter)")
ax1[1].axis("off")

plt.tight_layout()
plt.savefig("task1_box_filter.png", dpi=150, bbox_inches='tight')
plt.show()
print("Task 1 completed: 7x7 Box Filter applied.")


#task 2: Apply 5x5 and 21x21 Gaussian filters


# Apply 5x5 Gaussian filter (sigma chosen automatically based on kernel size)
gaussian_5x5 = cv2.GaussianBlur(image, (5, 5), 0)

# Apply 21x21 Gaussian filter
gaussian_21x21 = cv2.GaussianBlur(image, (21, 21), 0)

# Plot all three images side by side
fig2, ax2 = plt.subplots(1, 3, figsize=(18, 5))
fig2.suptitle("Task 2: Gaussian Filters (5x5 vs 21x21)", fontsize=16, fontweight='bold')

ax2[0].imshow(image)
ax2[0].set_title("Original Image")
ax2[0].axis("off")

ax2[1].imshow(gaussian_5x5)
ax2[1].set_title("5x5 Gaussian Filter")
ax2[1].axis("off")

ax2[2].imshow(gaussian_21x21)
ax2[2].set_title("21x21 Gaussian Filter")
ax2[2].axis("off")

plt.tight_layout()
plt.savefig("task2_gaussian_filter.png", dpi=150, bbox_inches='tight')
plt.show()
print("Task 2 completed: 5x5 and 21x21 Gaussian Filters applied.")



# Step 1: Read the image (already done above)
# Step 2: Convert image to float32
image_float = image.astype(np.float32) / 255.0

# Step 3: Apply the Laplacian filter using cv2.Laplacian()
laplacian = cv2.Laplacian(image_float, cv2.CV_32F, ksize=3)

# Step 4: Sharpen the image
sharpened = np.clip(image_float - laplacian, 0, 1)

# Step 5: Plot original, Laplacian result, and sharpened image
fig3, ax3 = plt.subplots(1, 3, figsize=(18, 5))
fig3.suptitle("Task 3: Laplacian Sharpening (3x3)", fontsize=16, fontweight='bold')

ax3[0].imshow(image_float)
ax3[0].set_title("Original Image")
ax3[0].axis("off")

# Normalize Laplacian for display (shift to 0-1 range for visibility)
laplacian_display = (laplacian - laplacian.min()) / (laplacian.max() - laplacian.min())
ax3[1].imshow(laplacian_display)
ax3[1].set_title("Laplacian Filter Output")
ax3[1].axis("off")

ax3[2].imshow(sharpened)
ax3[2].set_title("Sharpened Image")
ax3[2].axis("off")

plt.tight_layout()
plt.savefig("task3_laplacian_sharpening.png", dpi=150, bbox_inches='tight')
plt.show()
print("Task 3 completed: 3x3 Laplacian Sharpening applied.")

print("\nAll tasks completed successfully!")
```
